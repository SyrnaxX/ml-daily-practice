```python
import pandas as pd
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
data = pd.read_csv(r"C:\Users\LENOVO\Downloads\titanic.csv",usecols=['Age','Survived','Fare'])
```

```python
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data['Age_Mean'] = data['Age'].fillna(data['Age'].mean())
data['Age_Median'] = data['Age'].fillna(data['Age'].median())
```

```python
print(data['Age_Mean'].var())
print(data['Age_Median'].var())
print(data['Age'].var())
```

```python
# after replacing nan value with mean/median the varience is srink 
```

```python
import matplotlib.pyplot as plt
```

```python
fig , (ax1,ax2,ax3) = plt.subplots(ncols=3,figsize = (14,3))
fig = data['Age_Mean'].plot(kind='kde',ax= ax2)
ax1.set_title('Before Imputation')
fig = data['Age'].plot(kind='kde',ax = ax1)
ax2.set_title('After Imputation(mean)')
fig = data['Age_Median'].plot(kind='kde',ax=ax3)
ax3.set_title('Age Imputation(Median)')
```
