Step 2 : Data Wrangling/Munging
- It is the process of cleaning , transforming and organizing raw data into<br>a format that is suitable analysis,visualization or machine learning
- t includes
1) Gathering Data(csv,API,web scraping,DataBases)
2) Accessing Data.
- Finding the number of rows/columns(shape)
- Data types of various columns (info())
- checking for missing values (info() or isna().sum())
- checking duplicates (is_unique())
- High level mathematical overview of the data (discribe())
3) Cleaning Data
- Missing Data(mean,mode,midian)
- Removing Duplicates (drop_duplicates)
- incorrect data type (astype())
