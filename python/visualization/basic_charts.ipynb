# **Load Datasets**
## 1. Wine Dataset

```python
from sklearn.datasets import load_wine
import pandas as pd
wine = load_wine()
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame(wine.data,columns=wine.feature_names)
df['WineType'] = [wine.target_names[wine_typ] for wine_typ in wine.target]
df = df.dropna()
df.head()
```

# **sacatter plot**

```python
plt.figure(figsize=(15,4))
plt.scatter(x = df['alcohol'],y=df['malic_acid'],color = 'tomato',alpha=0.50,marker='s')
plt.xlim(10,16) # limits the y_labels
plt.ylim(0,7) # limits the y_labels
plt.xlabel('alcohol',fontdict= dict(fontsize = 11,fontweight = 'bold')) 
plt.ylabel('Malic acid',fontdict= dict(fontsize = 11,fontweight = 'bold'))
plt.title('Relationship between Alcohol & Malic_acid',fontdict= dict(fontsize=20,fontweight = 'bold'),loc='left')
```

```python
plt.figure(figsize=(15,4))

# for loop 
for i in df.WineType.unique():
    plt.scatter(x = df['alcohol'].where(df['WineType']==i),
                y = df['malic_acid'].where(df['WineType']==i) ,s=75,alpha=0.8,marker='s',label=i) # s : size

plt.xlim(10,16) # limits the y_labels
plt.ylim(0,7) # limits the y_labels
plt.xlabel('alcohol',fontdict= dict(fontsize = 11,fontweight = 'bold')) 
plt.ylabel('Malic acid',fontdict= dict(fontsize = 11,fontweight = 'bold'))
plt.legend(loc='best')
plt.title('Relationship between Alcohol & Malic_acid',fontdict= dict(fontsize=20,fontweight = 'bold'),loc='left')
```

# **Bar chart**

```python
avg_wine = df.groupby('WineType').mean().reset_index()
```

```python
avg_wine
```

```python
plt.figure(figsize=(20,4))
plt.bar(x=avg_wine['WineType'],height=avg_wine['flavanoids'],width=0.85,color='mediumseagreen',edgecolor='black',alpha=0.7,linewidth=2.0)
plt.xlabel('Wine Type',fontdict=dict(fontsize=12,fontweight='bold'))
plt.ylabel('Flavanoids',fontdict= dict(fontsize=12,fontweight='bold'))
plt.yticks(range(0,4))
plt.title('Avg Flavanoids per Wine Type',fontdict= dict(fontsize = 22,fontweight='bold'),pad=15,loc='left')
```

# **Pie Chart**

```python
df_count = df.groupby('WineType')['alcohol'].count().rename('Count').reset_index()
```

```python
plt.figure(figsize=(6,6))
plt.pie(x=df_count['Count'],labels= df_count['WineType'],colors=['tomato','mediumpurple','mediumseagreen'])
plt.title('Wine Type Dist.',fontdict=dict(fontsize = 22,fontweight='bold'),loc='left',pad=20)
plt.tight_layout()
plt.legend(loc='best')
```

```python
plt.figure(figsize=(6,6))
plt.pie(x=df_count['Count'],labels= df_count['WineType'],colors=['indianred','gold','hotpink'],autopct='%.2f%%',shadow=True,explode=[0,0.1,0])
plt.title('Wine Type Dist.',fontdict=dict(fontsize = 22,fontweight='bold'),loc='left',pad=20)
plt.tight_layout()
plt.legend(loc='best')
```

```python
plt.figure(figsize=(6,6))
plt.pie(x=df_count['Count'],labels= df_count['WineType'],colors=['tomato','mediumpurple','mediumseagreen'],autopct='%.2f%%',wedgeprops=dict(width=0.5),pctdistance=0.7,textprops=dict(fontsize=11,fontweight='bold',color='white'))
plt.title('Wine Type Dist.',fontdict=dict(fontsize = 22,fontweight='bold'),loc='left',pad=20)
plt.tight_layout()
plt.legend(loc='best')
```
