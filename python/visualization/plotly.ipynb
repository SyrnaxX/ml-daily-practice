## **Heatmap**
Used to find the patterns coorelation 

## **Correlation Between Travel Features**

```python
import plotly.figure_factory as ff
## Compute correlations matrix(numerical columns only)
correalation_matrix= df.select_dtypes(include=['number']).corr()

##Create annotated heatmap with smaller font
fig = ff.create_annotated_heatmap(
    z=correalation_matrix.values,
    x=correalation_matrix.columns.tolist(),
    y= correalation_matrix.index.to_list(),
    colorscale='Viridis',
    showscale= True,
    annotation_text=[[f'{val:2f}'for val in row]for row in correalation_matrix.values],
    font_colors = ["white"]
)
fig.show()
```

## Inferance:
- inverse relationship: higher rating(1.0) correlates with lower tourist arrivals(0.098) suggesting premium-rated hotels may attract fewer,possibly hogh-end travelers.
- Price vs Demand: The highest hotel price (1.0) has mid-range hotel booking (0.461) , indicating price senstivity- luxury stays may not always maximimize occupancy.
- Travel cost impact: Peak average travel cost (1.0) aligns with high hotel bookings(0.911), implying popular seasons.(e.g summer)drive both cost and demand.
- key insight Balance between affordability (price/bookings) and exclusivity (rating/arrivals) is critical in tourism trends.
