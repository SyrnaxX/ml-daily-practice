## **Array Attributes**

```python
a1 = np.arange(10 ,dtype='int32') # vector
a2 = np.arange(12,dtype=float).reshape(4,3) # matrix
a3 = np.arange(8).reshape(2,2,2) # Tensor
# int32 : takes 4 bit
# int64 : takes 8 bit
```

```python
# ndim (no. of dimensions)
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
# shape (shape of matrix(rows,columns))
print(a1.shape)
print(a2.shape)
print(a3.shape)
```

```python
# size (no of items)
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
# itemsize ( how much each item occupy space in memory)
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

```python
# dtype (data type of each item)
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

# **Array Operations**

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
```

```python
# scalar operations
print(a1+2)
print()
print(a1-2)
print()
print(2*a1)
print()
print(a2/2)
print()
print(a1**2)
```

```python
# relational
print(5>a1)
print()
print(9==a1)
```

```python
# Vector Operations
print(a1+a2)
print()
print(a1-a1)
print()
print(a1*a2)
print()
print(a1/a2)
```

# **Array Functions**

```python
# max/min/sum/prod(product)
print(np.max(a1)) 
print(np.min(a1,axis=0)) # get min  row wise (0: Columns , 1: Rows)
print(np.sum(a1))
print(np.prod(a1))
```

```python
# mean/median/std/var
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
```

```python
# trigonometry functions
np.sin(a1)
```

```python
#  dot product
a1 = np.arange(1,13).reshape(3,4)
a2 = np.arange(13,25).reshape(4,3)
print(np.dot(a1,a2))
```

```python
# log and exponents
print(np.log(a1))
print()
print(np.exp(a1))
```

```python
# round/floor/cell
print(np.round(np.random.random((2,3))*100)) # roundoff to nearest integer 
print()
print(np.floor(np.random.random((2,9)))) 
```

# **Indexing and Silicing**

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(4,3)
a3 = np.arange(8).reshape(2,2,2)
```

```python
print(a1[-1])
print(a2)
```

```python
# access these elements 6,11,9,4
print(a2[2,0])
print(a2[3,2])
print(a2[3,0])
print(a2[1,1])
```

```python
# access elements from 3d array
print(a3)
```

```python
# access elements (5,2,7,0,6)
print(a3[1,0,1])
print(a3[0,1,0])
print(a3[1,1,1])
print(a3[0,0,0])
print(a3[1,1,0])
```

```python
# Slicing
print(a1)
```

```python
# slicing in 1D
print(a1[2:5])
print(a1[5:9])
print(a1[::-3])
```

```python
# Slicing in 2D
print(a2)
```

```python
# access [3,4] [6,7]
print(a2[1:3,0:2])
print()
# access [0,2] ,[9,11]
print(a2[0::3,0::2])
# access [3,5] [9,11]
print()
print(a2[1::2,0::2])
# access [4],[7]
print()
print(a2[1:3,1:2])
```

```python
# Access elements from 3d array
a3 = np.arange(27).reshape(3,3,3)
print(a3)
```

```python
# access [9,10,11],[12,13,14] [15,16,17]
print(a3[1])
print()
# access first and last matrix
print(a3[::2])
print()
# access [[3,4,5]]
print(a3[0,1,:]) 
# access [10,13,16]
print()
print(a3[1,:,1])
# 22,23,35,26
print()
print(a3[2,1:,1:])
print()
# access 0,2,18,20
print(a3[::2,:6:6,::2])
print()
# access 6,8,24,26
print(a3[::2,2,::2])
```

## **Iterating**

```python
for i in a1:
    print(i,end=" ")
```

```python
for i in a2:
    print(i) # print 1d array (vector) 
```

```python
for i in a3:
    print(i,end= '\n\n')  #  print 2d array(Matrix)
```

```python
# for element by iteration 
for i in np.nditer(a3): # convert nd   array to 1d  array than print each element
    print(i,end = " ")
```

```python
for i in a3.flatten():
    print(i,end= ' ')
```

```python
for i in a3.ravel():
    print(i,end= ' ')
```

# **Stacking**
- Add two array

```python
# Horizontal stacking
a1 = np.arange(1,13).reshape(3,4)
a2 = np.arange(13,25).reshape(3,4)
np.hstack((a1,a2)) # pass in tuple
```

```python
# vertical stacking
np.vstack((a1,a2))
```

# **Splitting**

```python
# horizontal splitting [ | ]
np.hsplit(a1,2) # (array,parts)
```

```python
# vertical split []
#               ------
#                []
np.vsplit(a1,3)
```
