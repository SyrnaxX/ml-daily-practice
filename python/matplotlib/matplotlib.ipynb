# **Matplotlib: Data Visualization in Python**

```python
import numpy as np
import matplotlib.pyplot as plt 
```

## **Figure Class**
**Defination:**

The Figure is the overall container for all the elements in the Matplotlib plot

It acts as a canvs where multiple plots(Axes) can be placed.

**Application:**

Used when creating multiple subplots in a single figure.

Helpful for complex visualizatins with different charts in on display. 

```python
#Example:
fig , ax =plt.subplots()
ax.set_title("Basic Figure and Axes")
ax.set_xlabel("Days")
ax.set_ylabel("Total patients")
plt.show()
```

## **Axes Class**

**Defination:**

The Axes class represents individual plots inside a Figure.

Each figure can have multiple Axes, each containing a seprate graph.

**Application:**

Used whenplotting multiple releated visualizations together(e.g sales trends vs product categories in retain analytics).

Helps control labels,ticks,legends and other customization for each plot.




# **Line Plot**
A line plot displays data points conntected by lines. It is commonly used for showing trends over time.
## **Application:**
Used for time series analysis(e.g monthly sales,wedsite traffic,tempreture changes).

```python
months = ["Jan","Feb","Mar","Apr","May","Jun"]
recovery_rate = [65,70,60,78,80,85]
plt.plot(months,recovery_rate,marker ="o",label="Recovery_rate",color = "r")
plt.xlabel("Months")
plt.title("Recovery Rate")
plt.ylabel("Recovery Percentage")
plt.legend()
plt.show()
```

```python
# inferenace
# Trend of recovery percentage is upward from january,except for the month of March is downward
```

# **Bar Plot**
A bar chart represents categorical data using rectangular bars ,where the height represents a numerical value.
## **Application:**
used to compare different categories (e.g Sales per product customer segmentation.)

in healthcare, it can show the number of patients in different hospital departments.

# **Problem Statement:**
### Bar Plot: Medicine Demand Analysis
### A pharmacy wants to analyze the demand for different medicines.
### We use a bar plot to visualize the demand.

```python
medicines = ['Paracetamol','Antibiotics','Painkillers','Insulin',"Anatacids"]
demand = [500,700,450,300,650]
plt.bar(medicines,demand,color=['red','blue','green','purple','orange'])
plt.xlabel("Medicine Type")
plt.title("Medicine Demand Analysis")
plt.ylabel("Units Sold")
plt.show()
```

```python
# inferance
# highest selling is of Antabiotics around 700,
# lowest selling of insulin , around 300,rest are between this range of 300-700
```

# **Pie Chart**
Defination:

A pie chart shows proportions of different categories within a whole.

Application:

Used for percentage distribution(e.g market share of different brands).

In healthcare, it can visualize percentage distribution of deaseas cases(e.g diabities ,hypertension,heart disease.etc.)

# **Problem statement:**
Pie Chart: Disease Distribution

A healthcare centre analyzed the distribution of diseases in a city.

We viualize the distribution using pie chart. 

```python
diseases = ["Diabities","Hypertension","Heart Disease","Asthma","Others"]
percentage = [30,25,15,10,20]
colors =['blue','red','green','purple','orange']
plt.pie(percentage,labels=diseases,autopct='%1.1f%%',colors=colors,shadow=True,wedgeprops={"linewidth":10})
plt.title("Diseases Distribution in City")
plt.show()
```

```python
#inferance 
# Most of the peoples in city are suffuring from diabities(about 30%)
# least are asthma around 10% 
```

## Scatter Plot

A sactter plots shows individual data points on a 2D plane .It is used identify the releationships between two variables.

Application:

Used for correlation analysis(e.g advertising spend vs revenue).

In healthcare. It can show correlation between BMI and Blood Sugar Levels.

## **Problem Statement:**

## **Scatter plot Doctor patient interaction Time**
## A hospital examines how time spent by doctors affects diogonsis accuracy.
## We visualze this realitionship with scatter plot

```python
time_spent = [10,15,20,25,30,35,40]
diagonisis_accuaracy =[60,65,70,75,80,85,90]
plt.scatter(time_spent,diagonisis_accuaracy,color='red')
plt.xlabel("Time Spent by Docoter(minutes)")
plt.ylabel("Diagonisis Accuaracy(%)")
plt.title('Doctor-Patient Interaction Time vs Diagonsis Accuracy')
plt.grid()
plt.show()
```

# **Histrogram:**

**Defination:**

A histrogram displays the frequency of a dataset by grouping values into bins.

**Application:**

Used for distribution analysis(e.g age distribution of customers)

In healthcare used to analyze blood pressure distribution among patients

# Problem Statement:
#### **We analyze the age distribution of patients visiting a hospital using a histrogram.**

```python
age = np.random.randint(10,80,100)
plt.hist(age,bins=10,color="blue",edgecolor = 'black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid()
plt.show()
```

```python
# inferance 
# highest number of patients of age group 72-75 ,that is around 16,
# Lowest number of patients are in of age around 18-25 , 58-65.
```

## **Word Cloud**

Defination

A word cloud is a visual representation of text data Where size of each word is proptional to its frequency in the dataset.

Application:

Use for text analysis(e.g most common customer complaints).

In healthcare, it can analyze patient feedback to identify common concerns.

### **Word cloud:** Common Symtoms Analysis
## Problem Statement
A hospital collects common symptoms from patient records.
We visualize frequently  occuring symptoms ysing a word cloud.

```python
from wordcloud import WordCloud
symptoms_text = "fever cough cold headache nausea fatigue pain weekness dizziness allergy infectiom sore throught congestion"
wordcloud = WordCloud(width=800,height=400,background_color='White').generate(symptoms_text)
plt.figure(figsize=(10,5))
plt.imshow(wordcloud,interpolation= 'bilinear')
plt.axis("off")
plt.title("Common Symtoms Analysis")
plt.show() 

```
