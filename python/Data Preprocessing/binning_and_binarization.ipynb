## **Binning/Discretization**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split , cross_val_score
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
```

```python
from sklearn.preprocessing import KBinsDiscretizer
```

```python
titanic = pd.read_csv(r"C:\Users\LENOVO\Downloads\titanic.csv",usecols=['Age','Survived','Fare'])
```

```python
titanic.sample(5)
```

```python
titanic['Age'] =  titanic['Age'].fillna(titanic['Age'].mean())
```

```python
x = titanic.iloc[:,[1,-1]]
y = titanic.iloc[:,0]
```

```python
x_train , x_test , y_train , y_test = train_test_split(titanic.iloc[:,[1,-1]],titanic.iloc[:,0],test_size=0.2,random_state=42)
```

```python
x_test.shape,x_train.shape
```

```python
y_train.shape,y_test.shape
```

```python
dt = DecisionTreeClassifier()
```

```python
dt.fit(x_train,y_train)
```

```python
y_pred = dt.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(DecisionTreeClassifier(),x,y,cv= 10, scoring='accuracy'))
```

```python
kbin_age = KBinsDiscretizer(n_bins=10,encode='ordinal',strategy='quantile')
kbin_fare = KBinsDiscretizer(n_bins=10,encode='ordinal',strategy='quantile')
```

```python
trf = ColumnTransformer(transformers=[
    ('first',kbin_age,[0]),
    ('Second',kbin_fare,[1])
])
```

```python
x_train_trf = trf.fit_transform(x_train)
x_test_trf = trf.transform(x_test)
```

```python
trf.named_transformers_['first'].n_bins_
```

```python
trf.named_transformers_['first'].bin_edges_
```

```python
trf.named_transformers_['Second'].bin_edges_
```

```python
output = pd.DataFrame({
    'age': x_train['Age'],
    'age_trf' : x_train_trf[:,0],
    'fare': x_train['Fare'],
    'fare_trf': x_train_trf[:,1]
})
```

```python
output
```

```python
clf = DecisionTreeClassifier()
clf.fit(x_train_trf,y_train)
y_pred2 = clf.predict(x_test_trf)
```

```python
accuracy_score(y_test,y_pred2)
```

```python
x_trf = trf.fit_transform(x)
np.mean(cross_val_score(DecisionTreeClassifier(),x,y,cv=10,scoring= 'accuracy'))
```

# **Binarization**

```python
titanic = pd.read_csv(r"C:\Users\LENOVO\Downloads\titanic.csv",usecols=['Age','Survived','Parch','Fare','SibSp'])
```

```python
titanic.head()
```

```python
titanic['family'] = titanic['SibSp'] + titanic['Parch']
```

```python
titanic.head(3)
```

```python
titanic.drop(columns=['SibSp','Parch'],inplace=True)
```

```python
titanic.head()
```

```python
titanic.dropna(inplace=True)
```

```python
x = titanic.iloc[:,1:]
y =  titanic[['Survived']]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# without binarization
clf = DecisionTreeClassifier()
clf.fit(x_train,y_train)
y_pred = clf.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(DecisionTreeClassifier(),x,y,cv=10,scoring='accuracy'))
```

```python
from sklearn.preprocessing import Binarizer
```

```python
# applying binarization
trf = ColumnTransformer([
    ('bin',Binarizer(copy=False),['family'])
],remainder='passthrough')
```

```python
x_train
```

```python
x_train_trf = trf.fit_transform(x_train)
x_test_trf = trf.transform(x_test)
```

```python
pd.DataFrame(x_train_trf,columns=['family','Age','Fare'])
```

```python
clf1 = DecisionTreeClassifier()
clf.fit(x_train_trf,y_train)
y_pred2 = clf.predict(x_test_trf)
accuracy_score(y_test,y_pred2)
```

```python
x_trf = trf.fit_transform(x)
np.mean(cross_val_score(DecisionTreeClassifier(),x_trf,y,cv=10,scoring='accuracy'))
```
