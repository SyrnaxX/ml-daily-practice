```python
import pandas as pd
import numpy as np
```

```python
dt = {'Age':[25,34,87,89,7],
      'Cabin': ['C23','B43','A22','D35','A11'],
      'Noise': [2,3,'A','Z','A']}
data = pd.DataFrame(dt)
```

```python
data
```

```python
data['Noise'].unique()
```

```python
fig = data['Noise'].value_counts().plot(kind='bar')
fig.set_title('Bar Plot')
```

```python
# extract numerical plot
data['Noise_numeric'] = pd.to_numeric(data['Noise'],errors='coerce',downcast='integer')
```

```python
data['Noise_categorical'] = np.where(data['Noise_numeric'].isna(),data['Noise'],np.nan)
```

```python
data.head()
```

```python
data['Cabin_num'] = data['Cabin'].str[1:] # captures numerical part
data['Cabin_cat'] = data['Cabin'].str[0]
```
