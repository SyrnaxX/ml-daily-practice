## **Handling Date And Time**

```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv(r"C:\Users\LENOVO\Downloads\aapl_stock_data.csv",usecols=['Date'])
```

```python
data.head()
```

```python
data.info()
```

```python
data['Date'] = pd.to_datetime(data['Date'])
```

```python
data.info()
```

## **Extract Year**

```python
data['year'] = data['Date'].dt.year
data.head(3)
```

## **Extract Month**

```python
data['month_no'] = data['Date'].dt.month
data.sample(3)
```

```python
data['month_name'] = data['Date'].dt.month_name()
data.sample(3)
```

```python
data['Day'] = data['Date'].dt.day
data.sample(3)
```

```python
# day of week
data['DOW'] = data['Date'].dt.day_of_week
data.sample(3) 
```

```python
data['Day_name'] = data['Date'].dt.day_name()
data.sample(3) 
```

```python
# is weekend ?
data['is_weekend'] = np.where(data['Day_name'].isin(['Sunday','Saturday','Friday']),1,0)
data['is_weekend'].value_counts()
```

```python
# extract week of the year
data['Week'] = data['Date'].dt.weekday
data.head()
```

```python
# quarter
data['quater'] = data['Date'].dt.quarter
data.sample(3)
```

```python
# semester
data['semester'] = np.where(data['quater'].isin([1,2]),1,2)
data.sample(3)
```

```python
# time elapsed between dates
import datetime
today= datetime.datetime.today()
(today - data['Date']).dt.days
```
