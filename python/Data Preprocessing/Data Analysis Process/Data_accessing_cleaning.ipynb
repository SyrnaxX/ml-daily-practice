# Note - Accessing data is an iterative Process

## **Data Quality Dimensions**
- compleatness ---> is data missing?
- validity ---> is data invalid --> negative height --> duplicated_patient_id
- Accuracy --> data is valid but not accurate ---> weight is 1 kg
- consistency ----> both valid and accurate but written differntly ---> new york ans NY

compleatness <---validity<---accuracy<---consistency

# **Data Cleaning Order**
- Quality --> compleateness
- Tidiness
- Quality --> validity
- Quality ----> accuracy
- Quality ---> consistency

```python
# both dataset tratments and treatment_cut column(hb1ac_change is difference of hba1c_start and hba1c_end)
```

```python
treatments['hba1c_change'] =  treatments['hba1c_start'] - treatments['hba1c_end']
```

```python
treatments.info()
```

```python
treatment_cut['hba1c_change'] =  treatment_cut['hba1c_start'] - treatment_cut['hba1c_end']
```

```python
treatment_cut.info()
```

```python
# in patients column contact has 
```

```python
treatment_df = pd.concat([treatments,treatment_cut],ignore_index=True)
```

```python
treatment_df = treatment_df.melt(id_vars=['given_name','surname','hba1c_start','hba1c_end','hba1c_change'],var_name='Type',value_name='dose_range')
```

```python
treatment_df = treatment_df[treatment_df['dose_range'] != '-']
```

```python
treatment_df['dose_range'] = treatment_df['dose_range'].str.replace('u','')
```

```python
treatment_df
```

```python
treatment_df['dose_start'] = treatment_df['dose_range'].str.split(' ').str.get(0)
```

```python
treatment_df['dose_end'] = treatment_df['dose_range'].str.split(' ').str.get(2)
```

```python
treatment_df  =treatment_df.drop(columns='dose_range')
```

```python
treatment_df[['dose_start','dose_end']] = treatment_df[['dose_start','dose_end']].astype('int16')
```

```python
treatment_df.reset_index(drop=True)
```
