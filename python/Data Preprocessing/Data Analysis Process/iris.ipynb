```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(20,4))
sns.kdeplot(df['petal_length'],label = 'petal length',fill=True)
sns.kdeplot(df['petal_width'],label = 'petal width',fill=True)
plt.title("Distribution of Petal length and Petal width ",fontdict= dict(fontsize=22,fontweight='bold'),loc='left',pad=20)
plt.legend()
```

```python
plt.figure(figsize=(20,4))
sns.kdeplot(df['sepal_length'],label = 'sepal length',fill=True)
sns.kdeplot(df['sepal_width'],label = 'sepal width',fill=True)
plt.title("Distribution of Sepal length and Sepal width ",fontdict= dict(fontsize=22,fontweight='bold'),loc='left',pad=20)
plt.legend()
```
