```python
tranformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(strategy='mean'),['age']),
    ('tnf2',OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']]),['review','education']),
    ('tnf3',OneHotEncoder(drop='first'),['gender'])
],remainder = 'passthrough')
```
