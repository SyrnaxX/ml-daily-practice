# **Predicting Sales Based on Store Footfalls using Simple Linear Regression**

## **Importing required libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.DataFrame({"Footfalls":[10,15,18,25,33,35,40],
                     "DailySales":[5000,7800,11000,12500,16000,17500,20000]})
data
```

## **Define independent (Footfalls) and dependent (DailySales) variables**

```python
x=data[['Footfalls']]
y=data[['DailySales']]
```

```python
# double brackets for 2d(rows and columns),single are use for series
```

```python
plt.xlabel('Footfalls')
plt.ylabel('Daily Sales')
plt.scatter(x,y,color='red',marker='o')
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
#Fit is a method for creating model or tranning datavor creating relations between x and y.
```

```python
print(f"cofficient (slope): {model.coef_}")
print(f"Intercept: {model.intercept_}")
```

simple linear regression eq: y= B0+B1*x

**(1) Predict sale of with footfall of 20**

```python
model.intercept_ + model.coef_*20
```

```python
model.intercept_ + model.coef_*35 # footfall of 35
```

## **Predict sales using the trained model for new footfall values**

```python
#Prepare new footfall balues (for Example: 45,50,60)
new_footfalls = pd.DataFrame({"Footfalls":[45,50,60]})
new_footfalls
```

```python
# Predict new Sales
```

```python
predicted_sales = model.predict(new_footfalls)
print("Predicted Daily Sales for new Footfalls:\n\n")
df1 =pd.DataFrame({'PREDICTED_SALES':predicted_sales.flatten()})
```

```python
new_df = pd.DataFrame({"Footfalls": new_footfalls['Footfalls'],'Predicted_Footfalls': predicted_sales.flatten()})
new_df
```
