## <center>**Linear Regression Class**<center/>

```python
class linear_regression:
    def __init__(self):
        self.m = None
        self.b = None

    def fit(self,X_train,Y_train):
       num = 0
       den = 0
       for i in range(X_train.shape[0]):
           num += ((X_train.iloc[i] - X_train.mean())*(Y_train.iloc[i]-Y_train.mean()))
           den += (X_train.iloc[i]-X_train.mean())**2

       self.m = num/den # slope
       self.b = (Y_train.mean() - self.m*X_train.mean()) # intercept
       print('m',self.m)
       print('b',self.b)        
    def predict(self,X_test):
        y_pred = self.m*X_test + self.b
        return y_pred

```
